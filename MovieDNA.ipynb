```python
df = df.drop_duplicates(subset='id')
```

```python
df.loc[:, 'release_date'] = pd.to_datetime(df['release_date'], errors='coerce')
```

```python
df.loc[:, 'year'] = df['release_date'].dt.year
df.loc[:, 'month'] = df['release_date'].dt.month_name()
```

```python

df = df.copy()

df.loc[:, 'overview'] = df['overview'].fillna('')

df.loc[:, 'genre_metadata'] = df['genre_ids'].apply(lambda x: ' '.join(map(str, x)))

df.loc[:, 'soup'] = (df['overview'] + " " + df['genre_metadata']).str.lower()

def clean_text(text):
    return re.sub(r'[^a-zA-Z0-9\s]', '', str(text))

df.loc[:, 'soup'] = df['soup'].apply(clean_text)


tfidf = TfidfVectorizer(stop_words='english')


tfidf_matrix = tfidf.fit_transform(df['soup'])


cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

def clean_input(text):
    """Helper to clean user input so 'Avatar!' matches 'avatar'"""
    return re.sub(r'[^a-zA-Z0-9\s]', '', str(text)).lower().strip()

def recommend_expert(movie_list):
    valid_indices = []
    missing_movies = []

   
    df_clean_titles = df['title'].apply(clean_input)
    
    
    for movie in movie_list:
        clean_name = clean_input(movie)
        match = df[df_clean_titles == clean_name]
        
        if not match.empty:
            valid_indices.append(match.index[0])
        else:
            missing_movies.append(movie)

    
    if missing_movies:
        print(f" Note: These were not in the trending list: {', '.join(missing_movies)}")

    if not valid_indices:
        return " I couldn't find any of those movies to base a recommendation on."

    blended_scores = cosine_sim[valid_indices].mean(axis=0)
    
    sim_scores = list(enumerate(blended_scores))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    final_indices = []
    for i, score in sim_scores:
        if i not in valid_indices:
            final_indices.append(i)
        if len(final_indices) == 6:
            break

    print(f"\n Predictions based on common features of {len(valid_indices)} movies:")
    return df[['title', 'vote_average', 'release_date']].iloc[final_indices]


```

```python
my_selections = ["Avatar", "Project Hail Mary", "Apex", "The Crow",'Michael',"Lee Cronin's The Mummy"] 

results = recommend_expert(my_selections)
print(results)
```
